Q: create a bar graph on th details student_csv

```python
import pandas as pd
```

```python
df=pd.read_csv("student_marks.csv")
print(df.info())
print(df.drop_duplicates(inplace=True))
colors=["g","b","k","c"]
plt.bar(df.Name,df.Maths,
        width=0.5,
        edgecolor="r",
        color=colors)
plt.xticks(rotation=45)
plt.yticks(rotation=30)
plt.tight_layout()
plt.show()
```

horizontal graph

```python
x=[12,4,8,18]
y=[10,6,4,16]
colors=["g","b","k","c"]
plt.barh(x,y,
        height=0.9,
        color=colors,
        edgecolor="r",
        alpha=0.5)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

pie chart

```python
y=[12,34,80,42,58]
plt.pie(y)
plt.show()
```

styles in pie chart

```python
y=[12,34,80,42,58]
label=["physics","chemistry","maths","biology","history"]
colors=["g","b","k","c","r"]
plt.pie(y,
        labels=label,
        colors=colors)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
y=[12,34,80,42,58]
label=["physics","chemistry","maths","biology","history"]
colors=["g","b","k","c","r"]
ex=[0.5,0.0,0.0,0.0,0.0]
plt.pie(y,
        labels=label,
        colors=colors,
        explode=ex)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
y=[12,34,80,42,58]
label=["physics","chemistry","maths","biology","history"]
colors=["g","b","y","c","r"]
ex=[0.5,0.0,0.1,0.0,0.2]
plt.pie(y,
        labels=label,
        colors=colors,
        explode=ex,
        autopct="%2.2f%%",
        labeldistance=0.3,
        startangle=45,
        shadow=True,
        )
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

3D-Plots

```python
fig=plt.figure()
ac=fig.add_subplot(111,projection="3d")
x=[1,2,3,4,5]
y=[1,2,3,4,5]
z=[0,0,0,0,0]
dx=[0.5,0.5,0.5,0.5,0.5]
dy=[0.5,0.5,0.5,0.5,0.5]
dz=[10,20,30,40,50]
ac.bar3d(x,y,z,dx,dy,dz,
        color="r")
plt.show()
```
