```python
import matplotlib.pyplot as plt
from numpy import random
```

```python
markers=["o","*","s","d"]
linestyle=[":","--","-.","-"]
colors=["r","g","b","k"]
y=[]
fig,axes=plt.subplots(4,4,figsize=(8,8))
for i in range(4):
    for j in range(4):
        x=random.randint(1,99,size=5)
        axes[i,j].plot(x,
                    marker=markers[i],
                    linestyle=linestyle[i],
                    c=colors[i]
                    )
        axes[i,j].set_title(f"plot Number{i},{j}")
plt.tight_layout()
plt.show()
```
