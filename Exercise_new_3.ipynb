### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

```python
df=pd.read_csv("u.user",sep="|")
df.head()
```

### Step 3. Assign it to a variable called users.

```python
users=df
users.head()
users.info()
users.columns
users.axes
```

```python
users.head(10)
```

### Step 4. Discover what is the mean age per occupation

```python
user_group_occ=users.groupby("occupation").age.mean().reset_index()

user_group_occ

```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
# users_male=users[users.gender=="M"]
# users_male=users_male[["occupation","gender"]]
users["Male"]=users.gender=="M"
users.groupby("occupation")["Male"].mean().sort_values(ascending=False).reset_index(name="male_ratio")
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
users.groupby("occupation")["age"].agg(["min","max"]).reset_index()
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
user_grp_gender_occ=users.groupby(["occupation","gender"]).age.mean().reset_index(name="mean_age")
user_grp_gender_occ=pd.DataFrame(user_grp_gender_occ)
user_grp_gender_occ
```

### Step 8.  For each occupation present the percentage of women and men

```python
users["Male"]=users["gender"]=="M"
users["Female"]=users["gender"]=="F"
users
# total_age=users["age"].sum()
# print(total_age)
users["Male"]=users["gender"]=="M"
users["Female"]=users["gender"]=="F"
user_male_female_per=users.groupby("occupation")[["Male","Female"]].mean()*100
user_male_female_per
```
