```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
```

fetch data

```python
df=pd.read_csv("profit_and_loss.csv")
df.head()
```

exploring and cleaning data


```python
df.info()
```

```python
df.columns
```

plotting graph for the column profit and loss

```python
profit=df[df["Profit/Loss"]>0]
profit.rename(columns={"Profit/Loss":"Profit"},inplace=True )
loss=df[df["Profit/Loss"]<0]
loss.rename(columns={"Profit/Loss":"Loss"},inplace=True )
print(profit)
print(loss)
plt.plot(profit["Profit"],
         marker="*",
         linestyle=":",
         c="#750202",
         label="Profit")
plt.plot(loss["Loss"],
         marker="o",
         linestyle="--",
         c="#1167B8",
         label="Loss")
plt.legend(loc="upper left")
plt.show()
```
