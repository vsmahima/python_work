```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_and_loss.csv")
print(df.head())
df.info()
```

```python
year=df["Year"]
profit_loss=df["Profit/Loss"]
colors=[]
for i in profit_loss:
    
    if(i>0):
        colors.append("g")
    else:
        colors.append("r")
plt.bar(year,profit_loss,
        color=colors)
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.xlabel("year")
plt.ylabel("profit/loss")
plt.tight_layout()
plt.show()
# colors
```
