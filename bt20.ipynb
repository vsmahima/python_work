question: add new column total marks containing the sum of 6 subject and a column avg calculate the avg or mean

```python
df=pd.read_csv("college_students.csv")
df.head(5)
df.columns
df.drop(["Attendance (%)","Semester"],axis=1,inplace=True)

df["Total_marks"]=df.select_dtypes(include=["number"]).sum(axis=1)
df["Average MArks"]=df.select_dtypes(include=["number"]).drop("Total_marks",axis=1).mean(axis=1)
df.head(5)

# drop column grade and re calculate grade from Average marks
df.drop(["Grade"],axis=1,inplace=True)
value=[]
for i in df["Average MArks"]:
    if(i>90):
        value.append("A")
    elif(i>80):
        value.append("B")
    elif(i>70):
        value.append("C")
    elif(i>60):
        value.append("D")
    elif(i>50):
        value.append("E")
    else:
        value.append("F")
df["Grade"]=value
df.head(15)

```
